```python
SAVE_DIR = "./../data/random/"
```

# EMA

```python
# load csv file
df_ema = pd.read_csv('./../data/EMA/EMA.csv')
df_ema.head(3)
```

```python
print("n unique medicine names:", df_ema.Name_of_medicine.nunique())
print("dataset length:", len(df_ema))
```

```python
print(Counter(df_ema.Name_of_medicine))
```

```python
# create tuples of "Name_of_medicine" and "EMA_product_number"
tuples = []
for i in range(len(df_ema)):
    tuples.append((df_ema.iloc[i]['Name_of_medicine'], df_ema.iloc[i]['EMA_product_number']))

tuples[:10]
```

```python
def save_random_tuples_to_file(tuples, filename):
    save_dir = os.path.dirname(filename)
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)
    # save tuples to file
    with open(filename, 'w') as f:
        for item in tuples:
            f.write(f"{item[0]}\t{item[1]}\n")

def get_random_tuples(tuples, n, seed=42):
    copy_tuples = tuples.copy()
    random.seed(seed) # set seed for reproducibility
    random.shuffle(copy_tuples)
    r_n = copy_tuples[:n]
    # count unique medicines
    unique_meds = set([item[0] for item in r_n])
    print(f"unique medicines in sample size {n}: {len(unique_meds)}")
    save_random_tuples_to_file(r_n, f'{SAVE_DIR}/EMA_{n}.txt')
    return r_n
```

```python
r_5 = get_random_tuples(tuples, 5)
r_10 = get_random_tuples(tuples, 10)
r_20 = get_random_tuples(tuples, 20)
r_50 = get_random_tuples(tuples, 50)
r_100 = get_random_tuples(tuples, 100)
r_150 = get_random_tuples(tuples, 150)

r_5 # inspect
```

# SwissPAR

```python
swisspar_dir = './../data/SwissPar/SwissPAR_Jan19_2025'
```

```python
def save_random_filenames_to_file(filenames, filename):
    save_dir = os.path.dirname(filename)
    if not os.path.exists(save_dir):
        os.makedirs(save_dir)
    with open(filename, 'w') as f:
        for item in filenames:
            f.write(f"{item}\n")

def get_random_filenames(dirname, n, seed=42):
    random.seed(seed) # set seed for reproducibility
    files = os.listdir('./data/SwissPar/SwissPAR_Jan19_2025')
    random.shuffle(files)
    r_n = files[:n]
    save_random_filenames_to_file(r_n, f'{SAVE_DIR}/SwissPar_{n}.txt')
    return r_n
```

```python
r_5 = get_random_filenames(swisspar_dir, 5)
r_10 = get_random_filenames(swisspar_dir, 10)
r_20 = get_random_filenames(swisspar_dir, 20)
r_50 = get_random_filenames(swisspar_dir, 50)
r_100 = get_random_filenames(swisspar_dir, 100)
r_150 = get_random_filenames(swisspar_dir, 150)

r_5
```

# Combine extracted data from 2 jsons

```python
import json
import sys
import pandas as pd
from pathlib import Path
```

```python
clean_json = "./../../inference/combined/FDA_manually_cleaned.json"
clean_df = pd.read_json(clean_json).transpose()
clean_df
```

```python
indications = "./../../data/FDA/with_extracted_data_disease_class/FDA.json"
indications_df = pd.read_json(indications).transpose()
indications_df = indications_df[["MA_Number", "Indications_and_usage_disease_class_extracted", "Indications_and_usage_disease_name_extracted"]]
indications_df.head()
```

```python
col_to_merge_by = "MA_Number"

merged_df = pd.merge(clean_df, indications_df, on=col_to_merge_by, how="outer")
merged_df = merged_df.rename(columns={
    "Indications_and_usage_disease_class_extracted": "Disease_class(es)",
    "Indications_and_usage_disease_name_extracted": "Disease_name(s)"
})
merged_df.head(-30)
```

```python
save_dir = Path("./../../inference/combined/")
# with open(save_dir / "FDA_final.json", "w") as f:
#     json.dump(merged_df.to_dict(), f)
merged_df.to_json(save_dir / "FDA_final.json", orient="index", indent=4)
```

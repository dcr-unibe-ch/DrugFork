```python
import pandas as pd
from pathlib import Path
```

```python
# Eingabe- und Ausgabepfad
input_file = Path("../../data/HealthCanada/with_extracted_data/HEALTHCANADA.csv")
output_file = input_file.with_name(input_file.stem + "_with_links.csv")

# Basis-URL
BASE_URL = "https://health-products.canada.ca/dpd-bdpp/info?lang=eng&code="

# CSV einlesen
df = pd.read_csv(input_file, dtype=str)  # dtype=str, um führende Nullen zu behalten

# Links generieren
df["dpd_info_url"] = BASE_URL + df["Marketing_authorisation_number"].str.strip()

# Neue CSV speichern
df.to_csv(output_file, index=False)

print(f"Fertig. Datei gespeichert unter: {output_file}")
```
